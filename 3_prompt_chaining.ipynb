```python
from langgraph.graph import StateGraph, START, END
from langchain_groq import ChatGroq
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatGroq(model = "llama-3.1-8b-instant")
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
    evaluate:int
```

```python
graph = StateGraph(BlogState)
```

```python
def create_outline(state: BlogState) -> BlogState:
    title = state['title']
    prompt = f'generate  a detailed outline for a blog on the topic - {title}'
    outline = model.invoke(prompt).content
    state['outline'] = outline

    return state
```

```python
def create_blog(state:BlogState) -> BlogState:
    title = state['title']
    outline = state['outline']

    prompt = f'write a detail blog on the title - {title} using the following outline \n{outline}'

    content = model.invoke(prompt).content

    state['content'] = content

    return state
```

```python
def evluate(state: BlogState) -> BlogState:
    outline = state['outline']
    content = state['content']
    prompt = f'based on the outline = {outline} rate my blog - {content}'
    evlaute = model.invoke(prompt).content
    state['evaluate'] = evlaute

    return state
    
    
```

```python
graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)
graph.add_node('evluate', evluate)
```

```python
graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog','evluate')
graph.add_edge('evluate',END)
```

```python
workflow = graph.compile()
```

```python
initial_state = {'title':'Rise of AI in India'}

final_state = workflow.invoke(initial_state)

print(final_state)
```
